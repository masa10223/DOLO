```python
import os,re,glob
```

```python
for path in glob.glob('../../video/orco_L1/*'):
    filename = os.path.splitext(os.path.basename(path))[0]
    script = f"nohup python3 drosophila_predict.py \
          --video_path ../video/orco_L1/{filename}.avi \
          --output_gif_path ../annotations/overfittings/gifs/{filename}.gif \
          --output_mov_path ../annotations/overfittings/movs/{filename}.mov  \
          --output_csv_path ./csvs/trajectory/{filename}.csv  \
          --start_frame 0 \
          --end_frame 3000 \
          --max_id 1 --dist_thresh 10 --iou_thr 0.3 --max_missing_frames 5 \
          --model_path ../annotations/overfittings/overfits_weights/C1_L5_ori/weights/best.pt\
          --overlap_thresh 5 --head_tail_jump_thresh 40 > ../annotations/overfittings/logs/predict_{filename}.out ;"
    print(script)
```
